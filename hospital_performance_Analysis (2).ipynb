```python
#Calculating IQR bounds for all four columns

outlier_cols = [
    "Length_of_Stay",
    "Total_Bill",
    "Insurance_Covered",
    "Patient_Payment"
]

bounds = {}

for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    bounds[col] = {"lower": lower, "upper": upper}

    print(f"\n{col}")
    print("Lower bound:", lower)
    print("Upper bound:", upper)
    print("Potential outliers:", 
          ((df[col] < lower) | (df[col] > upper)).sum())
```

```python
# Checking potential outliers in each numerical column
columns = [
    'Length_of_Stay',
    'Total_Bill',
    'Insurance_Covered',
    'Patient_Payment'
]

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\n{col}")
    print("Outlier count:", len(outliers))
    print(outliers[col].sort_values(ascending=False).head(10))
```

```python
# Checking for invalid negative values
for col in columns:
    print(f"{col}:")
    print("Negative values:", (df[col] < 0).sum())
    print("Minimum value:", df[col].min())
    print("Maximum value:", df[col].max())
```

```python
# Checking financial consistency
print("Insurance covered > Total bill:")
print((df['Insurance_Covered'] > df['Total_Bill']).sum())

print("\nPatient payment > Total bill:")
print((df['Patient_Payment'] > df['Total_Bill']).sum())

print("\nInsurance + patient payment > Total bill:")
print(
    (
        df['Insurance_Covered'] + df['Patient_Payment']
        > df['Total_Bill']
    ).sum()
)
```

```python
# Inspecting records where combined payments exceed the total bill
check = df[
    (df['Insurance_Covered'] + df['Patient_Payment'])
    > df['Total_Bill']
].copy()

check['Payment_Difference'] = (
    check['Insurance_Covered']
    + check['Patient_Payment']
    - check['Total_Bill']
)

print("Records to review:", len(check))

display(
    check[
        [
            'Patient_ID',
            'Total_Bill',
            'Insurance_Covered',
            'Patient_Payment',
            'Payment_Difference'
        ]
    ].head(10)
)
```

```python
# Allowing a small tolerance for floating-point differences
tolerance = 0.01

combined_payment = (
    df['Insurance_Covered'] + df['Patient_Payment']
)

print(
    "Records exceeding bill by more than 0.01:",
    (combined_payment > df['Total_Bill'] + tolerance).sum()
)

print(
    "Records where payments approximately equal bill:",
    (
        (combined_payment - df['Total_Bill']).abs()
        <= tolerance
    ).sum()
)
```
